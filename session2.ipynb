```python
plays = [120, 135, 150, 200, 120, 90, 200]

mean = sum(plays) / len(plays)

print("Mean number of plays:", mean)
```

```python
delivery_times = [30, 25, 40, 35, 30, 45, 30]

delivery_times.sort()

n = len(delivery_times)
median = delivery_times[n // 2]

print("Median delivery time:", median)
```

```python
def most_common_rating(ratings):
    return max(set(ratings), key=ratings.count)

ratings = [5, 4, 4, 3, 5, 4, 2, 4]

result = most_common_rating(ratings)

print("Most common rating:", result)
```

```python
import statistics

channel1 = [100, 120, 110, 105, 115]
channel2 = [200, 220, 210, 205, 215]
channel3 = [100, 120, 110, 105, 5000]

channels = [channel1, channel2, channel3]

for i in range(len(channels)):
    data = channels[i]

    print("Channel", i + 1)
    print("Mean:", statistics.mean(data))
    print("Median:", statistics.median(data))

    try:
        print("Mode:", statistics.mode(data))
    except statistics.StatisticsError:
        print("Mode: No unique mode")

    print()
```

```python
import pandas as pd
data = {
    "Runs": [45, 60, 30, 60, 55]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("Mean:", df["Runs"].mean())
print("Median:", df["Runs"].median())
print("Mode:", df["Runs"].mode().tolist())
```
