```python
def bayes_posterior(prior, likelihood, evidence):
    posterior = (prior * likelihood) / evidence
    return posterior


prior = 0.4
likelihood = 0.7
evidence = 0.5

result = bayes_posterior(prior, likelihood, evidence)

print("Posterior Probability:", result)
```

```python
# Prior probabilities
prior_premium = 0.10
prior_regular = 0.90

# Probability of offering a discount
likelihood_premium = 0.80
likelihood_regular = 0.20

# Total probability of seeing a discount
evidence = (prior_premium * likelihood_premium) + (prior_regular * likelihood_regular)

# Bayes theorem
posterior = (prior_premium * likelihood_premium) / evidence

print("Probability that the restaurant is premium:", round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")
```

```python
# List of 20 email subjects
emails = [
    "Free gift for you", "Meeting tomorrow", "Win free money",
    "Project report", "Free offer now", "Lunch at noon",
    "Claim your free prize", "Class schedule", "Important notice",
    "Free vacation", "Assignment submission", "Team meeting",
    "Free cash reward", "Your order confirmed", "Hello friend",
    "Free entry contest", "Exam timetable", "Weekly report",
    "Free shopping voucher", "See you soon"
]

# Email labels: spam or not spam
labels = [
    "spam", "not spam", "spam", "not spam", "spam",
    "not spam", "spam", "not spam", "not spam", "spam",
    "not spam", "not spam", "spam", "not spam", "not spam",
    "spam", "not spam", "not spam", "spam", "not spam"
]

total_emails = len(emails)
spam_count = labels.count("spam")
not_spam_count = labels.count("not spam")

# Count emails containing the word 'free'
spam_free = sum(
    1 for email, label in zip(emails, labels)
    if label == "spam" and "free" in email.lower()
)

not_spam_free = sum(
    1 for email, label in zip(emails, labels)
    if label == "not spam" and "free" in email.lower()
)

# Calculate probabilities
prior_spam = spam_count / total_emails
likelihood_free_spam = spam_free / spam_count
likelihood_free_not_spam = not_spam_free / not_spam_count

evidence = (
    prior_spam * likelihood_free_spam
    + (not_spam_count / total_emails) * likelihood_free_not_spam
)

posterior = (prior_spam * likelihood_free_spam) / evidence

print("Spam emails containing 'free':", spam_free)
print("Non-spam emails containing 'free':", not_spam_free)
print("Probability email is spam given 'free':", round(posterior, 2))
print("Percentage:", round(posterior * 100, 2), "%")
```

```python
# Prior probabilities
prior_phone = 0.30
prior_no_phone = 0.70

# Likelihood probabilities
likelihood_headphones_phone = 0.60
likelihood_headphones_no_phone = 0.10

# Total probability of buying headphones
evidence = (
    prior_phone * likelihood_headphones_phone
    + prior_no_phone * likelihood_headphones_no_phone
)

# Bayes theorem
posterior = (
    prior_phone * likelihood_headphones_phone
) / evidence

print("Probability customer bought a phone:", round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")
```
