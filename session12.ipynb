```python
matrix1 = [
    [10, 20],
    [15, 25]
]

matrix2 = [
    [2, 3],
    [4, 5]
]

result = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += matrix1[i][k] * matrix2[k][j]

print("Resulting Matrix:")

for row in result:
    print(row)
```
