```python
print("Null Hypothesis (H0): People spend equal or less time on Instagram than YouTube.")

print("Alternative Hypothesis (H1): People spend more time on Instagram than YouTube.")
```

```python
from scipy import stats
import math

mean_A = 350
mean_B = 320
std_A = 50
std_B = 50
n_A = 30
n_B = 30

t_value = (mean_A - mean_B) / math.sqrt(
    (std_A**2 / n_A) + (std_B**2 / n_B)
)

p_value = 2 * stats.t.sf(abs(t_value), df=n_A + n_B - 2)

print("T-value:", round(t_value, 2))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
from scipy import stats

old_clicks = 480
new_clicks = 520
total_users = 1000

old_rate = old_clicks / total_users
new_rate = new_clicks / total_users

print("Old Button Click Rate:", old_rate * 100, "%")
print("New Button Click Rate:", new_rate * 100, "%")

# Two-proportion z-test
pooled_rate = (old_clicks + new_clicks) / (2 * total_users)

standard_error = (
    pooled_rate * (1 - pooled_rate) * (2 / total_users)
) ** 0.5

z_value = (new_rate - old_rate) / standard_error

p_value = 2 * stats.norm.sf(abs(z_value))

print("Z-value:", round(z_value, 2))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("The difference is statistically significant.")
else:
    print("The difference is not statistically significant.")
```
