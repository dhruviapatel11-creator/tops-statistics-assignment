```python
ratings = [
    [5, 4, 3],
    [4, 5, 2],
    [3, 4, 5]
]

print("Ratings Matrix:")

for row in ratings:
    print(row)

print("Ratings given to the second food item:")

for i in range(3):
    print("Friend", i + 1, ":", ratings[i][1])
```

```python
orders = [
    [10, 15, 20, 12],
    [12, 18, 15, 10],
    [8, 20, 17, 15]
]

dish_totals = [0, 0, 0, 0]

for day in orders:
    for i in range(4):
        dish_totals[i] += day[i]

for i in range(4):
    print("Dish", i + 1, "total orders:", dish_totals[i])
```

```python
products = [
    [500, 10],
    [1000, 5],
    [250, 20],
    [750, 8],
    [1200, 4]
]

def calculate_revenue(price, quantity):
    return price * quantity

for i in range(len(products)):
    price = products[i][0]
    quantity = products[i][1]

    revenue = calculate_revenue(price, quantity)

    print("Product", i + 1, "revenue: ₹", revenue)
```
