```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7]
posts = [1, 2, 2, 3, 4, 4, 5]

plt.scatter(hours, posts)

plt.xlabel("Hours Spent on Instagram")
plt.ylabel("Number of Posts")
plt.title("Instagram Usage vs Number of Posts")

plt.show()

print("The graph shows a positive correlation.")
```

```python
import numpy as np

zomato_orders = [5, 8, 3, 10, 6, 9]
swiggy_orders = [4, 7, 2, 9, 5, 8]

correlation = np.corrcoef(zomato_orders, swiggy_orders)[0, 1]

print("Correlation coefficient:", round(correlation, 2))

if correlation > 0:
    print("Positive correlation")
elif correlation < 0:
    print("Negative correlation")
else:
    print("Zero correlation")
```
