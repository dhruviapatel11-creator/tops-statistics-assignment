```python
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

probability = calculate_probability(1, 2)

print("Probability of getting Heads:", probability)
```

```python
import random

count = 0

for i in range(100):
    dice = random.randint(1, 6)

    if dice == 6:
        count += 1

probability = count / 100

print("Number of times 6 appeared:", count)
print("Experimental probability:", probability)
```

```python
genres = [
    'pop', 'rock', 'pop', 'hiphop', 'pop',
    'jazz', 'rock', 'pop', 'hiphop', 'rock',
    'pop', 'jazz', 'pop', 'rock', 'hiphop',
    'pop', 'jazz', 'rock', 'pop', 'hiphop'
]

pop_count = genres.count('pop')
total_songs = len(genres)

probability = pop_count / total_songs

print("Number of Pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of selecting a Pop song:", probability)
```

```python
def calculate_joint_probability(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set.intersection(dessert_set)
    all_users = food_set.union(dessert_set)

    probability = len(both_users) / len(all_users)

    return probability


food_users = [1, 2, 3, 4, 5, 6]
dessert_users = [2, 3, 5, 7]

probability = calculate_joint_probability(food_users, dessert_users)

print("Probability of ordering both food and dessert:", probability)
```
