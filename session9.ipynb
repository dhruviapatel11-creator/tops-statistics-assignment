```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 100)

plt.hist(data, bins=10, density=True, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Density")
plt.show()

print("Mean:", round(np.mean(data), 2))
print("Standard Deviation:", round(np.std(data), 2))
```

```python
import numpy as np
from collections import Counter

steps = [5000, 6000, 7000, 8000, 7500,
         6000, 9000, 8500, 7000, 6500,
         8000, 7500, 6000, 7000, 9000,
         5500, 8000, 7500, 7000, 6500]

mean = np.mean(steps)
median = np.median(steps)

counts = Counter(steps)
max_count = max(counts.values())
mode = [value for value, count in counts.items()
        if count == max_count]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)

if abs(mean - median) < 1000:
    print("Mean and median are approximately equal.")
else:
    print("Mean and median are not approximately equal.")
```

```python
import numpy as np

scores = np.random.normal(70, 15, 200)

lower_limit = 70 - 15
upper_limit = 70 + 15

count = np.sum((scores >= lower_limit) & (scores <= upper_limit))

print("Total players:", len(scores))
print("Mean score:", round(np.mean(scores), 2))
print("Players scoring between 55 and 85:", count)
print("Percentage:", round(count / len(scores) * 100, 2), "%")
print("Expected percentage is approximately 68%.")
```
